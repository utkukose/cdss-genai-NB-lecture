## 1. Kartın doldurulması · Filling the canvas


```python
CANVAS = {
    'karar': (
        'Yoğun bakıma kabul edilen bir hastanın uzun süre kalıp kalmayacağının erken '
        'öngörülmesi. Sistem kabulden altı saat sonra devreye girer ve yatak kapasitesi '
        'planlaması ile erken yükseltme kararına girdi sağlar.'
    ),
    'kullanici': (
        'Yoğun bakım sorumlu hekimi ve yatak yönetiminden sorumlu hemşire. Sabah viziti '
        'sırasında, hasta listesi gözden geçirilirken.'
    ),
    'veri_turu': (
        'Rutin hastane verisi. Demografik bilgi, yatış bağlamı ve ilk altı saatin vital '
        'bulguları ile laboratuvar sonuçları.'
    ),
    'sonuc_degiskeni': (
        'Yoğun bakım yatış süresinin üç günü aşması. Yatış ve çıkış zamanlarından '
        'hesaplanır, rutin kayıtlarda doğrudan mevcuttur.'
    ),
    'prevalans': (
        'Bu kohortta yaklaşık üçte bir. Kendi biriminizde farklı olacaktır.'
    ),
    'pahali_hata': (
        'Kaçırma. Uzun kalacak bir hastanın öngörülmemesi kapasitenin plansız dolmasına '
        've başka bir hastanın kabulünün gecikmesine yol açar. Yanlış alarmın maliyeti '
        'gereksiz bir planlama toplantısıdır.'
    ),
    'yanilirsa': (
        'Yanlış negatifte kapasite planı yetersiz kalır ve devir kararı gecikir. '
        'Devretme kuralı: Sistem çekimser kaldığında ya da hasta eğitim popülasyonuna '
        'benzemediğinde karar sorumlu hekime geçer, sistem çıktısı gösterilmez.'
    ),
}

BASLIKLAR = {
    'karar': '1. Hangi klinik karar',
    'kullanici': '2. Kim kullanacak',
    'veri_turu': '3. Hangi veri türü',
    'sonuc_degiskeni': '4. Sonuç değişkeni',
    'prevalans': '5. Prevalans',
    'pahali_hata': '6. Hangi hata daha pahalı',
    'yanilirsa': '7. Sistem yanılırsa ne olur',
}

def kart_metni(canvas=CANVAS):
    bos = [k for k, v in canvas.items() if not str(v).strip()]
    if bos:
        print('Boş bırakılan sorular:', ', '.join(BASLIKLAR[k] for k in bos))
        print('Boş bırakmak yerine "bilmiyorum, şu sebeple" yazınız.')
        print()
    return '\n\n'.join(f'{BASLIKLAR[k]}: {canvas[k]}' for k in BASLIKLAR)

print(kart_metni())
```

## 2. Birinci istemin kurulması · Assembling prompt 1

Aşağıdaki hücre kartı birinci istemin içine yerleştirmekte ve kopyalanmaya hazır tam
metni yazdırmaktadır. Türkçe ve İngilizce sürümlerin ikisi de üretilmektedir.


```python
ISTEM_TR = '''Bir klinik karar destek sistemi tasarlamama yardım ediyorsun. Klinik bağlamım şu:

{kart}

Yalnızca aşağıdaki bölümlerden oluşan bir belirtim belgesi üret, başka hiçbir şey yazma:

- Karar anı: Sistemin çıktı ürettiği tam an ve o anda kullanıcının ne yapmakta olduğu.
- Tahmin hedefi: Sonucun kesin tanımı, zaman penceresi ve rutin kayıtlarda nasıl
  tespit edileceği.
- Girdiler: Karar anında gerçekten erişilebilir olan değişkenler ile ancak sonradan
  erişilebilecek değişkenler ayrı ayrı. Bu ayrımda katı ol.
- Çalışma noktası: Belirttiğim prevalans ve kaçırma ile yanlış alarm arasındaki maliyet
  farkı göz önüne alındığında duyarlılık mı yoksa özgüllük mü öncelenmeli? Tek
  paragrafta gerekçelendir.
- Başarısızlık biçimleri: Bu sistemin bir hastaya zarar verebileceği üç somut yol.
- Kapsam dışı: Bu sistemin açıkça yapmadığı şeyler.

Cevaplarımın eksik ya da kendi içinde tutarsız olduğu yerlerde boşluğu kendin
doldurma; bunu söyle. Açık kalan soruları en sonda listele.'''

print(ISTEM_TR.format(kart=kart_metni()))
```

```python
ISTEM_EN = '''You are helping design a clinical decision support system. Here is my clinical context:

{kart}

Produce a specification document with these sections, and nothing else:

- Decision point: the exact moment in the clinical workflow at which the system
  produces output, and what the user is doing at that moment.
- Prediction target: a precise definition of the outcome, including the time window
  and how it would be identified in routine records.
- Inputs: the variables that would realistically be available at the decision point,
  separated from variables that would only be available afterwards. Be strict about
  this separation.
- Operating point: given the prevalence I stated and the relative cost of a miss
  versus a false alarm that I stated, state which of sensitivity or specificity should
  be favoured and why, in one paragraph.
- Failure modes: three specific ways this system could harm a patient.
- Out of scope: what this system explicitly does not do.

Where my answers are incomplete or internally inconsistent, say so instead of filling
the gap yourself. List the open questions at the end.'''

print(ISTEM_EN.format(kart=kart_metni()))
```

## 3. Cevabın okunması · Reading the reply

Asistan bir belirtim üretecektir. Belgenin gövdesinden önce **en sondaki açık sorular
listesine** bakınız. O liste kartınızın zayıf noktalarıdır ve kuracağınız sistemin de
zayıf noktaları olacaktır.

Üç kontrol yapınız.

**Girdiler bölümü.** Karar anında mevcut olan ile sonradan mevcut olan değişkenleri
gerçekten ayırmış mı? Ayırmadıysa, bu sistemin hedef sızıntısı ile kurulacağını şimdiden
biliyorsunuz demektir.

**Çalışma noktası.** Sizin belirttiğiniz maliyet asimetrisine uygun yönde mi karar
vermiş? Kendi varsayılanına mı kaymış?

**Açık sorular listesi.** Boşsa şüphelenin. Yedi cevaptan oluşan bir kartın hiçbir
boşluk bırakmaması olağandışıdır; liste boşsa asistan büyük olasılıkla boşlukları
kendisi doldurmuştur.
