## Adım 2 · Değerlendirme

Bir modelin başarımı tek sayıyla anlatılamaz.

**Doğruluk yanıltır.** Hedef durum hastaların onda birinde görülüyorsa, herkese olumsuz
diyen bir kural yüzde 90 doğruluk verir. Doğruluk burada modelin ne yaptığını değil,
durumun ne kadar seyrek olduğunu ölçer.

**Ayrım gücü tek başına yetmez.** Modelin hastaları doğru sıralaması ile verdiği
olasılığın gerçeğe karşılık gelmesi ayrı şeylerdir. Eşik belirleyecekseniz ikincisi daha
önemlidir.

**Asıl soru klinik karşılıktır.** Yüz hastada kaç uyarı çıkıyor ve kaçı doğru? Bu satır,
16 Eylül dersindeki Epic Sepsis Model örneğinin sizin modelinizdeki karşılığıdır.


### İstem

```
Modelin başarımını ölçen bir fonksiyon yaz. Adı degerlendir olsun; gerçek sonuçları,
tahmin olasılıklarını ve bir eşik değerini alsın.

Şunları hesaplasın ve her birinin altına sade bir dille ne anlama geldiğini yazsın:

1. ROC eğrisi altındaki alan, güven aralığıyla birlikte
2. Kalibrasyon: Modelin verdiği olasılık gerçeğe ne kadar karşılık geliyor
3. Verilen eşikte duyarlılık, özgüllük, pozitif ve negatif kestirim değerleri
4. Klinik karşılık: Bu eşikte yüz hastada kaç uyarı çıkar, kaçı doğrudur, kaç vaka
   kaçırılır
5. Alt grup dökümü: Aynı ölçümleri demografik gruplara göre ayrı ayrı ver. Bir grupta
   vaka sayısı çok azsa sayı üretme, yetersiz örneklem yaz
6. Hiçbir şey öğrenmeyen bir kuralın, yani herkese aynı cevabı veren kuralın aynı
   ölçütlerdeki değerleri

Sonuçları ekrana yazdır ve bir sözlük olarak döndür. Fonksiyonu test kümesi üzerinde 0.50
eşiğiyle çalıştır, sonucu sonuc değişkeninde tut.
```
