## Adım 1 · Model genel olarak neye bakıyor

Ölçme yöntemi basittir: Bir değişkenin değerleri kayıtlar arasında rastgele karıştırılır
ve modelin başarımının ne kadar düştüğüne bakılır. Çok düşüyorsa model o değişkene
dayanıyor demektir.

Sonuçta yayılım, ortalama kadar önemlidir. Küçük bir kümede tekrarlar arasındaki fark
komşu değişkenler arasındaki farktan büyük olabilir; o zaman sıralamanın kendisi
güvenilir değildir.


### İstem

```
Modelin hangi değişkenlere dayandığını ölçen bir fonksiyon yaz. Adı genel_onem olsun.

Her değişken için değerleri kayıtlar arasında rastgele karıştır ve modelin başarımının ne
kadar düştüğünü ölç. Karıştırmayı yirmi kez tekrarla ki sonucun ne kadar oynadığı da
görünsün.

Sonucu bir veri çerçevesi olarak döndür: degisken, etki, yayilim sütunları bulunsun,
etkiye göre büyükten küçüğe sıralı olsun. Fonksiyonu test kümesi üzerinde çalıştır,
sonucu onem değişkeninde tut ve ilk on beş satırı göster.
```

- En etkili değişkenler hangileri? Klinik olarak beklediğiniz değişkenler mi?

- Yüksek etkili değişkenlerden biri klinik bir işaret yerine kaydın yapılış biçiminden

  sinyalde cihaz gürültüsü, metinde tekrarlayan şablon cümle böyle davranır. Hiçbiri
  klinik bulgu değil, bakım sürecinin izidir.

## Adım 2 · Tek hastada kararı ne belirledi

Genel açıklama modeli anlatır; tek hasta açıklaması o hastanın neden önünüze geldiğini
anlatır. Klinisyenin ihtiyaç duyduğu ikincisidir.

Üç hastaya bakacaksınız: Doğru yakalanan, boşuna uyarılan ve kaçırılan. **Asıl önemli
olan boşuna uyarılandır.** Yanlış bir tahmini makul gösteren bir açıklama,
açıklanabilirliğin hatayı akladığı yerdir.


### İstem

```
Tek bir hasta için modelin kararının gerekçesini üreten bir fonksiyon yaz. Adı
```
