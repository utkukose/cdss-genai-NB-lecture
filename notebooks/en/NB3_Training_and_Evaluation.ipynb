## Step 2 · Evaluation

The performance of a model cannot be conveyed by one number.

**Accuracy misleads.** If the condition occurs in one patient in ten, a rule that says
negative to everyone is ninety percent accurate. Accuracy there measures how rare the
condition is.

**Discrimination alone is not enough.** Ranking patients correctly and returning a
probability that corresponds to reality are different things.

**The real question is the clinical translation.** How many alerts per hundred patients
and how many are true? That line is the counterpart, in your model, of the Epic Sepsis
Model example from the lecture.
